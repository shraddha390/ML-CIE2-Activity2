## Activity 2 – Simple House Price Prediction Using Linear Regression
Dataset: California Housing. Input feature X: `median_income`. Target y: `median_house_value`. The activity requires an 80:20 split, Linear Regression, coefficient/intercept/equation, test predictions, MAE/MSE/R² (RMSE optional), actual-vs-predicted plot, and one new prediction.

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Load California Housing dataset
housing_url = "https://raw.githubusercontent.com/ageron/handson-ml2/master/datasets/housing/housing.csv"
housing = pd.read_csv(housing_url)
print("Shape:", housing.shape)
display(housing.head())
```

```python
# Inspect dataset
print("Columns:", housing.columns.tolist())
housing.info()
display(housing.describe().T)
display(housing.isnull().sum().to_frame("Missing Values"))
```

```python
# Select one input feature and one target; remove missing values in these columns
data = housing[["median_income", "median_house_value"]].dropna()
X = data[["median_income"]]
y = data["median_house_value"]
print("Input feature X: median_income")
print("Target y: median_house_value")
print("Usable rows:", len(data))
```

```python
# Scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(X["median_income"], y, alpha=0.3)
plt.title("Median Income vs Median House Value")
plt.xlabel("Median Income")
plt.ylabel("Median House Value")
plt.show()
```

```python
# 80% training data and 20% testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
# Train Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)
print("Coefficient (slope):", model.coef_[0])
print("Intercept:", model.intercept_)
print(f"Equation: Predicted House Value = {model.coef_[0]:.2f} * Median Income + {model.intercept_:.2f}")
```

```python
# Predict test-set values
y_pred = model.predict(X_test)
comparison = pd.DataFrame({
    "Actual House Value": y_test.values,
    "Predicted House Value": y_pred
})
display(comparison.head(10))
```

```python
# Evaluate model
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
print("MAE:", round(mae, 2))
print("MSE:", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("R-squared (R2):", round(r2, 4))
```

```python
# Actual vs predicted plot with perfect-prediction reference line
plt.figure(figsize=(7, 6))
plt.scatter(y_test, y_pred, alpha=0.3, label="Predictions")
low = min(y_test.min(), y_pred.min())
high = max(y_test.max(), y_pred.max())
plt.plot([low, high], [low, high], linestyle="--", label="Perfect prediction")
plt.title("Actual vs Predicted House Values")
plt.xlabel("Actual House Value")
plt.ylabel("Predicted House Value")
plt.legend()
plt.show()
```

```python
# Predict for a new example: median_income = 5.0 (dataset units)
new_house = pd.DataFrame({"median_income": [5.0]})
new_prediction = model.predict(new_house)
print("New input median_income:", 5.0)
print("Predicted median house value:", round(new_prediction[0], 2))
```

### Activity 2 – Conclusion
A Simple Linear Regression model was trained using `median_income` to predict `median_house_value`. The coefficient, intercept, equation, test predictions, evaluation metrics, actual-vs-predicted graph and a new prediction were generated.

**Limitation:** Only one feature is used. Actual house prices can depend on many other factors, so this simple model may not predict every house price accurately.
